# Face Authentication - Model Training
This notebook prepares and saves the face embedding model using DeepFace + FaceNet.

```python
import os
import numpy as np
import cv2
from deepface import DeepFace
import pickle
import json
```

```python
MODEL_NAME = "Facenet"
DETECTOR_BACKEND = "opencv"
THRESHOLD = 0.70
DISTANCE_METRIC = "cosine"
```

```python
def build_and_warm_model():
    print("Loading FaceNet model weights...")
    model = DeepFace.build_model(MODEL_NAME)
    print("Model loaded successfully.")
    return model
```

```python
def get_embedding(img_path, model):
    result = DeepFace.represent(
        img_path=img_path,
        model_name=MODEL_NAME,
        detector_backend=DETECTOR_BACKEND,
        enforce_detection=True
    )
    return np.array(result[0]["embedding"])
```

```python
def cosine_similarity(vec1, vec2):
    dot = np.dot(vec1, vec2)
    norm = np.linalg.norm(vec1) * np.linalg.norm(vec2)
    if norm == 0:
        return 0.0
    return dot / norm
```

```python
def verify_faces(img1_path, img2_path, model):
    emb1 = get_embedding(img1_path, model)
    emb2 = get_embedding(img2_path, model)
    
    score = cosine_similarity(emb1, emb2)
    
    label = "same person" if score >= THRESHOLD else "different person"
    
    return {
        "verification_result": label,
        "similarity_score": round(float(score), 4)
    }
```

```python
model = build_and_warm_model()
```

```python
config = {
    "model_name": MODEL_NAME,
    "detector_backend": DETECTOR_BACKEND,
    "threshold": THRESHOLD,
    "distance_metric": DISTANCE_METRIC
}

with open("model_config.json", "w") as f:
    json.dump(config, f, indent=4)

print("Model config saved to model_config.json")
print("Training setup complete. The model weights are managed by DeepFace internally.")
```
